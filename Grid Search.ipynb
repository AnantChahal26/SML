```python
import numpy as np
from scipy.stats import loguniform
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from sklearn.svm import SVC

train = np.loadtxt("Datasets/pendigits.tra", delimiter=",")
test = np.loadtxt("Datasets/pendigits.tes", delimiter=",")
Xtr, ytr = train[:, :16] / 100, train[:, 16].astype(int)
Xte, yte = test[:, :16] / 100, test[:, 16].astype(int)

grid = GridSearchCV(SVC(), {"C": [1, 10, 100], "gamma": [0.1, 1]}, cv=3)
grid.fit(Xtr, ytr)

print("Grid Search")
print("Combinations tried:", len(grid.cv_results_["params"]))
print("Best C =", grid.best_params_["C"], " Best gamma =", grid.best_params_["gamma"])
print("Best CV Accuracy:", round(grid.best_score_, 4))
print("Test Accuracy:", round(grid.score(Xte, yte), 4))

rand = RandomizedSearchCV(SVC(), {"C": loguniform(1, 100), "gamma": loguniform(0.1, 1)}, n_iter=6, cv=3, random_state=42)
rand.fit(Xtr, ytr)

print("Random Search")
print("Combinations tried:", len(rand.cv_results_["params"]))
print("Best C =", round(rand.best_params_["C"], 4), " Best gamma =", round(rand.best_params_["gamma"], 4))
print("Best CV Accuracy:", round(rand.best_score_, 4))
print("Test Accuracy:", round(rand.score(Xte, yte), 4))
```
