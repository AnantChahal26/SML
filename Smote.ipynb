```python
import numpy as np
import pandas as pd
from imblearn.over_sampling import SMOTE
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("Datasets/bank-full.csv", sep=";")
X = df[["age", "balance", "campaign", "pdays", "previous"]]
y = (df["y"] == "yes").astype(int)

Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

scaler = StandardScaler()
Xtr = scaler.fit_transform(Xtr)
Xte = scaler.transform(Xte)

print("Class counts before SMOTE [no, yes]:", np.bincount(ytr).tolist())

sm = SMOTE(random_state=42)
Xres, yres = sm.fit_resample(Xtr, ytr)

print("Class counts after SMOTE [no, yes]:", np.bincount(yres).tolist())

model = LogisticRegression(max_iter=500)
model.fit(Xres, yres)
prob = model.predict_proba(Xte)[:, 1]

print("Test ROC-AUC:", round(roc_auc_score(yte, prob), 4))
```
