```python
import numpy as np

np.random.seed(1)
n = 6
start = (0, 0)
goal = (5, 5)
walls = [(1, 2), (2, 2), (3, 2), (4, 4)]
actions = [(-1, 0), (1, 0), (0, -1), (0, 1)]
arrows = ["U", "D", "L", "R"]

alpha = 0.2
gamma = 0.95
eps = 0.2
episodes = 2500

Q = np.zeros((n, n, 4))


def step(s, a):
    r = s[0] + actions[a][0]
    c = s[1] + actions[a][1]
    if r < 0 or r >= n or c < 0 or c >= n or (r, c) in walls:
        return s
    return (r, c)


for ep in range(episodes):
    s = start
    for t in range(100):
        if np.random.rand() < eps:
            a = np.random.randint(4)
        else:
            a = np.argmax(Q[s[0], s[1]])
        s2 = step(s, a)
        if s2 == goal:
            target = 20
        else:
            target = -1 + gamma * Q[s2[0], s2[1]].max()
        Q[s[0], s[1], a] += alpha * (target - Q[s[0], s[1], a])
        s = s2
        if s == goal:
            break

print("Learned policy (W = wall, G = goal):")
for r in range(n):
    row = ""
    for c in range(n):
        if (r, c) in walls:
            row = row + "W "
        elif (r, c) == goal:
            row = row + "G "
        else:
            row = row + arrows[np.argmax(Q[r, c])] + " "
    print(row)

path = [start]
s = start
while s != goal and len(path) < n * n:
    s = step(s, np.argmax(Q[s[0], s[1]]))
    path.append(s)

print("Learned path:", path)
print("Number of moves:", len(path) - 1)
```
